````python
%%writefile app.py
import streamlit as st
import google.generativeai as genai
import json

st.set_page_config(page_title="ระบบแต่งนิยายและสตอรี่บอร์ด AI", layout="wide")
st.title("📖 ระบบแต่งนิยาย & สตอรี่บอร์ดภาพอัตโนมัติ")

with st.sidebar:
    st.header("⚙️ การตั้งค่าระบบ")
    api_key = st.text_input("ใส่ Gemini API Key:", type="password")
    art_style = st.selectbox(
        "เลือกลายเส้นภาพสตอรี่บอร์ด:",
        [
            "2D อนิเมะญี่ปุ่นคมชัด (Anime Style)",
            "ภาพสีน้ำแฟนตาซี (Watercolor Fantasy)",
            "ภาพวาดคอนเซ็ปต์อาร์ตภาพยนตร์ (Cinematic Concept Art)",
            "ดาร์กแฟนตาซี ลายเส้นหมึกดำ (Dark Fantasy Gothic)"
        ]
    )
    if api_key:
        genai.configure(api_key=api_key.strip())

col1, col2 = st.columns(2)
with col1:
    st.subheader("1. ข้อมูลตัวละครและโลก")
    characters = st.text_area("รายละเอียดตัวละคร:", height=150, placeholder="ชื่อ, หน้าตา, นิสัย...")
    world_setting = st.text_input("ธีม/เซ็ตติ้งของโลก:", placeholder="เช่น โตเกียวแนวอาถรรพ์...")

with col2:
    st.subheader("2. คอนเซ็ปต์เนื้อเรื่อง")
    story_concept = st.text_area("พล็อตเรื่องหรือเหตุการณ์:", height=150, placeholder="โครงเรื่องที่อยากให้เกิดขึ้น...")
    target_scenes = st.slider("จำนวนฉากสำคัญที่ต้องการสกัด:", min_value=10, max_value=30, value=20)

if st.button("🚀 เริ่มแต่งนิยายและสกัด 20 ฉากสำคัญ", type="primary"):
    if not api_key:
        st.error("กรุณาใส่ API Key ที่แถบด้านซ้ายก่อนครับ")
    elif not characters or not story_concept:
        st.warning("กรุณากรอกข้อมูลตัวละครและคอนเซ็ปต์ให้ครบถ้วน")
    else:
        with st.spinner("AI กำลังแต่งนิยายและคัดเลือกฉากสำคัญ..."):
            try:
                # กำหนดโมเดลรุ่นล่าสุดตามที่ API แนะนำ
                model = genai.GenerativeModel("models/gemini-3.8-flash")

                prompt = f"""
                คุณคือนักเขียนนิยายและผู้กำกับสตอรี่บอร์ดมืออาชีพ
                ธีมโลก: {world_setting}
                ตัวละคร: {characters}
                พล็อตเรื่อง: {story_concept}

                คำสั่ง:
                1. แต่งเนื้อเรื่องฉบับเต็มโดยคงบุคลิกตัวละครอย่างเคร่งครัด
                2. คัดเลือก "ฉากสำคัญ (Key Scenes)" จำนวน {target_scenes} ฉากตามลำดับเวลาตั้งแต่ต้นจนจบเรื่อง
                3. แปลงแต่ละฉากเป็นคำสั่ง Prompt ภาษาอังกฤษสำหรับส่งให้ AI วาดภาพ ระบุหน้าตาเสื้อผ้าให้สอดคล้องกันตลอด

                ตอบกลับเป็นโครงสร้าง JSON ตามนี้เท่านั้น (ห้ามใส่คำเกริ่นใดๆ):
                {{
                    "novel_text": "เนื้อเรื่องนิยายภาษาไทย...",
                    "scenes": [
                        {{
                            "scene_no": 1,
                            "title_th": "ชื่อฉากสั้นๆ",
                            "description_th": "บรรยายเหตุการณ์ในฉากภาษาไทย",
                            "image_prompt": "English visual description for image generation..."
                        }}
                    ]
                }}
                """

                res = model.generate_content(prompt)
                clean_text = res.text.replace("```json", "").replace("```", "").strip()
                data = json.loads(clean_text)
                st.session_state["data"] = data
            except Exception as e:
                st.error(f"เกิดข้อผิดพลาดในการประมวลผล: {e}")

if "data" in st.session_state:
    data = st.session_state["data"]
    tab1, tab2 = st.tabs(["📖 เนื้อเรื่องนิยาย", "🎬 บอร์ดภาพเหตุการณ์สำคัญ (20+ ฉาก)"])
    with tab1:
        st.markdown(data["novel_text"])
    with tab2:
        st.write(f"พบเหตุการณ์สำคัญ **{len(data['scenes'])} ฉาก**:")
        grid = st.columns(2)
        for idx, sc in enumerate(data["scenes"]):
            with grid[idx % 2]:
                st.markdown(f"**ฉากที่ {sc['scene_no']}: {sc['title_th']}**")
                st.caption(sc["description_th"])
                full_prompt = f"{sc['image_prompt']}, {art_style}"
                st.code(full_prompt, language="text")
````
